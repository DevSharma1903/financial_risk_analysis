# Prepare SEC data for the academic amendment proxy

Run in a Colab GPU runtime if desired; this stage is data preparation and does not fit a model. Put `data/raw/`, `src/09_prepare_colab_dataset.py`, and `src/12_update_sec_quarters.py` under `MyDrive/financial-risk-detector/`. The SEC downloads require your real identifying User-Agent.


```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
import os, getpass, subprocess
ROOT = Path('/content/drive/MyDrive/financial-risk-detector')
for name in ['src/09_prepare_colab_dataset.py', 'src/12_update_sec_quarters.py']:
    if not (ROOT/name).is_file():
        raise FileNotFoundError(ROOT/name)
if not (ROOT/'data/raw/2025Q3/sub.txt').is_file():
    raise FileNotFoundError('Upload the existing raw quarters under MyDrive/financial-risk-detector/data/raw/')
```

```python
agent = getpass.getpass('SEC User-Agent (your name/organization and contact email): ').strip()
if '@' not in agent or ' ' not in agent:
    raise ValueError('The SEC download requires an identifying User-Agent with contact email.')
env = {**os.environ, 'SEC_USER_AGENT': agent}
subprocess.run(['python', str(ROOT/'src/12_update_sec_quarters.py')], check=True, env=env)
```

```python
subprocess.run(['python', str(ROOT/'src/09_prepare_colab_dataset.py')], check=True)
print('Prepared for the training notebook:', ROOT/'data/colab_amendment_proxy.csv.gz')
```

The output is an amendment proxy. It does not identify confirmed restatements. The preparation step excludes original filings without a complete 365-day follow-up window.
